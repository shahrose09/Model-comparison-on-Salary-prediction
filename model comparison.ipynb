```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score


df=pd.read_csv("salary_data.csv")
print(df.columns)

x=df[["experience", "education_level", "skill_score", "projects"]]
y=df["salary"]


scalar=StandardScaler()
x_scaled=scalar.fit_transform(x)
pca=PCA(n_components=2)
x_pca=pca.fit_transform(x_scaled)
print("original shape:", x.shape)
print("reduced shape:", x_pca.shape)
pca_df=pd.DataFrame(data=x_pca,
                     columns=["PC1", "PC2"])


x_train, x_test, y_train, y_test = train_test_split(x_pca, y, test_size=0.2, random_state=42)

dt_model=DecisionTreeClassifier( max_depth=3, random_state=42)
dt_model.fit(x_train, y_train)
dt_y_pred=dt_model.predict(x_test)
print("Decision Tree Accuracy:", accuracy_score(y_test, dt_y_pred))

rf_model=RandomForestClassifier(n_estimators=100, random_state=42, max_depth=3)
rf_model.fit(x_train, y_train)
rf_y_pred=rf_model.predict(x_test)
print("Random Forest Accuracy:", accuracy_score(y_test, rf_y_pred))


svm_model=SVC(kernel="linear", random_state=42, C=1.0)

svm_model.fit(x_train, y_train)
svm_y_pred=svm_model.predict(x_test)

print("Accuracy:", accuracy_score(y_test, svm_y_pred))

new_employee=pd.DataFrame({
    "experience": [6],
    "education_level": [3],
    "skill_score": [78],
    "projects": [6]
})
new_employee_scaled = scalar.transform(new_employee)
new_employee_pca = pca.transform(new_employee_scaled)
prediction=dt_model.predict(new_employee_pca)
print("\nPredictions:")

print("Decision Tree:",
      dt_model.predict(new_employee_pca))

print("Random Forest:",
      rf_model.predict(new_employee_pca))

print("SVM:",
      svm_model.predict(new_employee_pca))


if prediction[0]==1:
    print("High Salary")
else:
    print("Low Salary")


```
